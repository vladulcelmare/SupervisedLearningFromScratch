```python
##### Linear Regression #####
# It is one of the most well-known ML algorithms thanks to its simplicity
# How does it work? It assumes a linear relationship between the input and output, being able to "draw a line", hence the name linear regression
# It uses the infamous formula y = mx + b, but for vectors : pred = x * w.T + bias, where bias is the vertical shift, and w is the slope
# For us to compute loss, we must use the mean squared error, and another evaluation metric known as r2
# Basically, r2 is used to track how well the model explains the variance of data in comparison with a model that simply takes the mean of data and uses it as prediction for the next input
# The formula to compute r2 is : 1 - model err/baseline err
# Disclaimer, this is my first ever kaggle notebook, so I'm sorry for the weird code and me not respecting conventions (also my first ML model).
import numpy as np
import pandas as pd
import kagglehub

df = pd.read_csv("/kaggle/input/datasets/vedavyasv/usa-housing/USA_Housing.csv")

def MSE(prediction, target): # mean squared error, can also compute mean absolute error which instead of taking square difference it takes absolute difference
    return np.mean( (prediction - target)**2 )

class LR():
    """
    Linear Regression model, built by me without respecting classical conventions. Sorry.
    """
    def __init__(self, iter : int = 20000, lr : float = 1e-3):
        self.r2 = None # r2
        self.max_iter = iter # maximum iterations
        self.lr = lr # learning rate
        self.w = None # slope
        self.beta = None # shift
        self.loss = None # loss
        
    def fit(self, x, y):
        """
        fit() function. x = labels, y = targets.
        """
        self.w = np.random.normal(size=(5, 1)) # we have the weight matrix
        self.beta = 0 # the bias
        
        for _ in range(self.max_iter):
            if _ % 1000 == 0 and _ != 0:
                print(_, np.sqrt(self.loss), self.r2)
                
            pred_y = self.w.T @ x + self.beta # we apply formula to get prediction
            e = pred_y - y # to make the calculation easier to process
            
            self.loss = MSE(pred_y, y) # we compute mean sq err
            self.r2 = 1 - (np.sum( (y - pred_y)**2) )/( np.sum( (y - np.mean(y))**2) ) # we compute r2
            
            self.w = self.w - self.lr * np.mean(2 * e * x, axis=1, keepdims=True) # we update weight and bias using GD
            self.beta = self.beta - self.lr * np.mean(2 * e)

    def predict(self, x):
        """
        predict() function. After training, we only provide labels and then look at the prediction.
        """
        return self.w.T @ x + self.beta


model = LR()
new_df = df.drop(columns=["Address"]) # i get rid of address column
y = np.array(list(new_df['Price'])) # i want to predict prices
y = y.reshape(1, -1) # i make it (1, 5000) to be the same as predicted y

x = [None] * 5
for i, a in enumerate(new_df):
    if i == 5:
        break
    x[i] = list(new_df[a]) # we memorise first 5 columns as 5 features

x = np.array(x)
x_mean = x.mean(axis=1, keepdims=True) # we standardize as we have large numbers
x_std = x.std(axis=1, keepdims=True)
x = (x - x_mean) / x_std
print(x.shape)

train_y = y[:, :int(y.shape[1] * 0.8)]
test_y = y[:, int(y.shape[1] * 0.8):]

train_x = x[:, :int(x.shape[1] * 0.8)]
test_x = x[:, int(x.shape[1] * 0.8):]

model.fit(train_x, train_y)
pred = model.predict(test_x)

test_mse = MSE(pred, test_y)
test_rmse = np.sqrt(test_mse)

test_r2 = 1 - np.sum((test_y - pred)**2) / np.sum((test_y - np.mean(test_y))**2)

print("train RMSE:", np.sqrt(model.loss))
print("train R2:", model.r2)
print("test RMSE:", test_rmse)
print("test R2:", test_r2)
print("difference in %", abs(model.r2 - test_r2) / abs(test_r2))

##### Use cases #####
# -> predicting continuous real values such as temperature, prices, pressure etc.
# -> simple and fairly-precise
# -> useful when there is a resemblance of a linear relationship between I/O
```
